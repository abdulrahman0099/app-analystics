```python
# filling null values with avg number of os

avgos= data.groupby('app_version')['os'].unique()
avgos
```

```python
filled = data.dropna(subset=['os']).drop_duplicates('app_version').set_index('app_version')['os'].to_dict()
filled
```

```python
# 2. Sirf NaN values fill karo
data['os'] = data['os'].fillna(data['app_version'].map(filled))

data.isnull().sum()
```

```python
data[data['os'].isna()][['os', 'app_version']]
```

```python
print(data.loc[data['app_version'].isna(), 'app_version'].tolist())
```
